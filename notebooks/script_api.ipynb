## Script para baixar e consumir a API


```python
import os
import json
from dotenv import load_dotenv
import requests

# Carrega as variáveis do arquivo .env para o ambiente
load_dotenv()

# Pega a chave da API com segurança do .env
API_KEY = os.getenv("VISUAL_CROSSING_API_KEY")
LOCATION = "Juiz de Fora"
UNIT_GROUP = "metric"

# 1. Define o nome da pasta e garante que ela exista
PASTA_DESTINO = "dados"
os.makedirs(PASTA_DESTINO, exist_ok=True)

# Constrói a URL do Endpoint
url = f"https://weather.visualcrossing.com/VisualCrossingWebServices/rest/services/timeline/{LOCATION}?unitGroup={UNIT_GROUP}&key={API_KEY}&contentType=json"

try:
    print(f"Buscando dados meteorológicos para {LOCATION}...")
    response = requests.get(url)
    
    # Lança um erro caso o status HTTP não seja 200 (ex: chave inválida, limite excedido)
    response.raise_for_status()
    
    # Converte a resposta para dicionário Python (JSON)
    data = response.json()
    
    # 2. Cria o caminho completo do arquivo dentro da pasta "dados"
    filename = os.path.join(PASTA_DESTINO, "weather_juiz_de_fora.json")
    
    # Salvando os dados brutos no arquivo JSON (se já existir, ele será atualizado/sobrescrito)
    with open(filename, 'w', encoding='utf-8') as f:
        json.dump(data, f, ensure_ascii=False, indent=4)
        
    print(f"Sucesso! Os dados foram salvos/atualizados no arquivo '{filename}'.\n")
    
    # Exemplo de uso rápido: exibindo as condições atuais
    if 'currentConditions' in data:
        current = data['currentConditions']
        print(f"--- Clima Atual em {LOCATION} ---")
        print(f"Temperatura: {current.get('temp')}°C")
        print(f"Sensação térmica: {current.get('feelslike')}°C")
        print(f"Condições: {current.get('conditions')}")
        print(f"Umidade: {current.get('humidity')}%")

except requests.exceptions.HTTPError as http_err:
    print(f"Erro HTTP: {http_err}")
    print("Dica: Verifique se a sua chave de API está correta no arquivo .env.")
except Exception as err:
    print(f"Ocorreu um erro inesperado: {err}")
```
